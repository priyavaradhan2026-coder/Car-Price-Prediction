```python
df['brand'] = df['name'].str.split(' ').str[0]
df['brand'].nunique()
```

```python
df.drop(['name'],axis=1,inplace=True)
df
```

```python
df['owner'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder
order = [['Test Drive Car', 'First Owner', 'Second Owner', 'Third Owner', 'Fourth & Above Owner']]
ordinal_enc = OrdinalEncoder(categories=order,dtype=int)
df['owner_encoded'] = ordinal_enc.fit_transform(df[['owner']])
df = df.drop('owner', axis=1)
df
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_fuel = OneHotEncoder(sparse_output=False, dtype=int)
encoded_fuel = one_hot_fuel.fit_transform(df[['fuel']])
colname_fuel = one_hot_fuel.get_feature_names_out()
df1 = pd.DataFrame(encoded_fuel, columns=colname_fuel)
df = pd.concat([df, df1], axis=1)
df = df.drop('fuel', axis=1)
df
```

```python
one_hot_seller = OneHotEncoder(sparse_output=False, dtype=int)
encoded_seller = one_hot_seller.fit_transform(df[['seller_type']])
colname_seller = one_hot_seller.get_feature_names_out()
df2 = pd.DataFrame(encoded_seller, columns=colname_seller)
df = pd.concat([df, df2], axis=1)
df = df.drop('seller_type', axis=1)
df
```

```python
one_hot_trans = OneHotEncoder(sparse_output=False, dtype=int)
encoded_trans = one_hot_trans.fit_transform(df[['transmission']])
colname_trans = one_hot_trans.get_feature_names_out()
df3 = pd.DataFrame(encoded_trans, columns=colname_trans)
df = pd.concat([df, df3], axis=1)
df = df.drop('transmission', axis=1)
df
```

```python
one_hot_brand = OneHotEncoder(sparse_output=False, dtype=int)
encoded_brand = one_hot_brand.fit_transform(df[['brand']])
colname_brand = one_hot_brand.get_feature_names_out()
df4 = pd.DataFrame(encoded_brand, columns=colname_brand)
df = pd.concat([df, df4], axis=1)
df = df.drop('brand', axis=1)
df
```

```python
df.shape
df
```

```python
x = df.drop('selling_price', axis=1)
y = df['selling_price']
x
y
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
print('total_data:',df.shape[0],'rows')
print('xtrain:',xtrain.shape[0],'rows')
print('xtest:',xtest.shape[0],'rows')
print('ytrain:',ytrain.shape[0],'rows')
print('ytest:',ytest.shape[0],'rows')
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
xtrain = scaler.fit_transform(xtrain)
xtest = scaler.transform(xtest)
```

```python
model=LinearRegression()
model.fit(xtrain,ytrain)
```

```python
#training and testing data efficiency :
print('training efficiency:',model.score(xtrain,ytrain))
print('testing efficiency:',model.score(xtest,ytest))
```

```python
ypred=model.predict(xtest)
ypred
```

```python
#model evoluation:
print("r2_score:",r2_score(ytest,ypred))
print("MSE:",mean_squared_error(ytest,ypred))
print("MAE:",mean_absolute_error(ytest,ypred))
```

```python
# new car prediction (full pipeline):

new_car = pd.DataFrame(0, index=[0], columns=x.columns)

new_car['year'] = 2015
new_car['km_driven'] = 40000
new_car['owner_encoded'] = 1
new_car['fuel_Petrol'] = 1
new_car['seller_type_Individual'] = 1
new_car['transmission_Manual'] = 1
new_car['brand_Maruti'] = 1

new_car = new_car[x.columns]  # ensure column order matches training exactly

new_car_scaled = scaler.transform(new_car)
predicted_price = model.predict(new_car_scaled)

print("Predicted price:", predicted_price[0])
new_car
```
